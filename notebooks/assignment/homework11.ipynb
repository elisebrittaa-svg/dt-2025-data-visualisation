Task 1 - Reading, writing, splitting and printing words from the linked file and text.

```python
import urllib.request #this is needed to work with urls

text = urllib.request.urlopen("https://raw.githubusercontent.com/elisebrittaa-svg/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/assignment/homework11lyrics.txt").read().decode("utf-8") #opened the text file by the github raw link and made it into a readable format
print(text) #prints the text in the terminal, so I can see if I did everything correct
```

```python
words = text.split(" ") #used this code to split the text into words, by separating words by spaces

print("The first word of the song Bohemian Rhapsody by Queen is:", words[0]) #printed the first word

print("The song Bohemian Rhapsody by Queen begins with the words:", *words[0:5]) #printed the first 5 words

print("The song Bohemian Rhapsody ends with the words:", *words[-5:]) #printed the last 5 words

print("In the song Bohemian Rhapsody, there are", len(words), "words") #printed the amount of words
```

```python
from PIL import Image #Import, to work with images
import urllib.request #Library needed, to work with URL
```

```python
import numpy as np #in order to work with image masks (using the backround and text on top of it)
```

```python
from wordcloud import WordCloud #necessary to create word clouds
```

```python
img = urllib.request.urlretrieve("https://drive.google.com/uc?export=download&id=1fVPipiP8wMO2LIn72rige59tn2Go-MmP", "backgroundimage.png")
#https://drive.google.com/file/d/1fVPipiP8wMO2LIn72rige59tn2Go-MmP/view?usp=sharing
# i created a new variable for the background image, opened it via google drive link (using file ID)
Image.open("backgroundimage.png")
#with this, opened it here in the terminal to see if I did everything correct
```

```python
WordCloud(mask=np.array(Image.open("backgroundimage.png")), stopwords=["and", "then", "but", "the", "to", "on", "s", "a", "my", "put", "his", "I'd", "as", "it", "can", "can't", "get", "got", "of", "so", "if", "go", "you", "I'm", "not", "too", "let"], background_color=("beige")).generate_from_text(text).to_image()
#(mask=np.array(Image.open("backgroundimage.png")) is to get the first image as a background
#I mark down stopwords=,  for words that shall not be included in the cloud word, connecting words mostly
#this code creates the word cloud based on the singer's picture shape and gives it the background colour that I wanted, beige.
```
