Task 1 - Reading, writing, splitting and printing words from the linked file and text.

```python
import urllib.request

text = urllib.request.urlopen("https://raw.githubusercontent.com/elisebrittaa-svg/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/assignment/homework11lyrics.txt").read().decode("utf-8")
print(text)
```

```python
words = text.split(" ")

print("The song Bohemian Rhapsody by Queen begins with the words:", *words[0:5]) #printed the first 5 words

print("The song Bohemian Rhapsody ends with the words:", *words[-4:])

```
