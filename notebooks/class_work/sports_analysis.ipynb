BOx plot for each athlete to find the number of years of practice.

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Years_of_practice", legend=False)
plt.ylabel("Years")
plt.title("Years of practice per athlete")
plt.show

#legend=False võtab nurgast ära tulpade pealkirja "Years_of_practice"
#plt.ylabels() annab y teljele pealkirja
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.ylabel("Age")
plt.title("Age of Athletes")
plt.show
```

Age vs. Years of practice

plt.scatter (x, y) makes points for each pair of values
"s" contains controls for the size of each bubble
s=df["Years_of_practice"]*30 means: if a student has more years of practice their bubble is larger, *3o is scaling factor to make bubble big.

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*30, alpha=0.6)
plt.xlabel("Age")
plt.ylabel("Years")
plt.title("Age vs Years of practice")
plt.show
```

Average as per sport

```python
avg_age=df.groupby("Sport") ["Age"].mean()
avg_age.plot(kind="bar")
plt.title("Aveage age per sport")
plt.ylabel("Average age")
plt.show
```
