```python
import pandas as pd
df= pd.read_csv("competencies.csv")
df
```

```python
df.describe() # generate some statistics
```

```python
avg_hours=df.groupby("Student")["Hours_Studied"].mean(). sort_values(ascending=False)

#Bar chart
avg_hours.plot(kind="bar", x="Student", y="Hours_Studied")
#titles
```

```python
avg_grades=df.groupby("Student")["Grade"].mean(). sort_values(ascending=False)

#Bar chart
avg_grades.plot(kind="bar", x="Student", y="Grade")

#titles
```

```python
avg_hours=df.groupby("Student")["Hours_Studied"].mean(). sort_values(ascending=False)

#Bar chart
avg_hours.plot(kind="pie", x="Student", y="Hours_Studied")


#titles
```

```python
avg_grades=df.groupby("Student")["Grade"].mean(). sort_values(ascending=False)

#Bar chart
avg_grades.plot(kind="bar", x="Student", y="Grade")

#titles
```

```python
import matplotlib.pyplot as plt
avg_grades=df.groupby("Student")["Grade"].mean(). sort_values(ascending=False)


plt.figure(figsize=(8,8))
#Bar chart
avg_grades.plot(kind="pie", x="Student", y="Grade")

#titles
```

```python
#Does more studying equal better grades? (Check exceptions)

plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6)
plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.title("Grade vs. Hours Studied")
plt.show
```

```python
#Calculate quartiles and extremes
hours_min = df["Hours_Studied"].min()
hours_q1 = df["Hours_Studied"].quantile(0.25)
print("Minimum hours:", hours_min, "Q1:", hours_q1)


#boxplot graph
plt.figure(figsize=(8,6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.show
plt.title("How concictent are study patterns and performance?")
plt.ylabel("Value")
```
