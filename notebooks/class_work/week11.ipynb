```python
sentence = "racecarlevelfield"

#extract first 3 letters, letter indicators start from 0
#[start:end]
print("Print first 3 letters:", sentence[0:4])
print(sentence[:4])

#extract letters from 5-9
print(sentence[4:10])

#extract lettes from 5th end
print(sentence[4:])

#extract the last character
print(sentence[-1])

#extract last four characters
print(sentence[-4:])

#extract first letter of the substring starting from five
#sentence[5:]: this will extract thee substring starting from 5
#[0]: this will only take the first character of that substring
print(sentence[5:][0])

#extract first 5 letters of the substring starting from 5
print(sentence[5:][:5])

#Extract entire sentence
print(sentence[:])

#Extract every 2nd character
print(sentence[::2])
```

```python
sentence = "I was devastated when I opened the R-Kiosk shop door and saw the mega long queue for the hotdog orders."

words = sentence.split(" ") #splits the sentence into words based on spaces

print(words)

#extract the first word
print(words[0])

#extract the first 3 words
print(*words[0:3])

#extract the last 3 words
print(*words[-3:])

#count how many words there are
print("The number of words:", len(words)) #len = lenght




```
